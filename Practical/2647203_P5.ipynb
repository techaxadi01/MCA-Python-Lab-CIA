```python
import numpy as np
```

## Task 1: Array Creation and Dataset Preparation

```python
'''
Column Index Reference:
0: Patient ID, 1: Age, 2: Temp (°C), 3: Heart Rate (bpm), 4: SpO2 (%), 5: Systolic BP (mmHg), 6: Diastolic BP (mmHg), 7: Daily Dose (mg), 8: Hospital Stay (days)
'''

patients = np.array([
    [101, 45, 37.2, 82, 97, 125, 82, 20, 4],
    [102, 62, 38.1, 96, 94, 145, 90, 30, 7],
    [103, 35, 36.8, 72, 99, 118, 76, 10, 3],
    [104, 71, 39.0, 105, 91, 160, 95, 40, 12],
    [105, 54, 37.5, 88, 96, 135, 85, 25, 6],
    [106, 29, 36.6, 68, 98, 110, 72, 10, 2],
    [107, 66, 38.5, 101, 93, 150, 92, 35, 9],
    [108, 48, 37.0, 78, 97, 128, 80, 5, 5],
    [109, 76, 39.2, 110, 89, 170, 100, 50, 15],
    [110, 41, 36.9, 75, 98, 120, 78, 15, 4]
], dtype=float)

baseline_thresholds = np.zeros((1, 9))
alert_weights = np.ones(9)
time_intervals = np.arange(1, 11, 1)
vital_norm_scale = np.linspace(0, 1, 10)

print("Patient Dataset successfully created.")
print("Dataset:\n", patients)
```

## Task 2 – Array Attributes and Patient Data Exploration

```python
print("--- Array Attributes ---")
print("Shape of Dataset             :", patients.shape)
print("Number of Dimensions (ndim)  :", patients.ndim)
print("Total Number of Elements     :", patients.size)
print("Data Type (dtype)            :", patients.dtype)
print("Bytes per Element (itemsize) :", patients.itemsize, "bytes")
print("Total Memory Occupied        :", patients.nbytes, "bytes")
```

## Task 3 – Indexing and Slicing

```python
# Extract Patient IDs (Column 0)
patient_ids = patients[:, 0].astype(int)
print("Patient IDs:", patient_ids)

# Extract Vital-Sign Columns: Temp, HR, SpO2, Systolic, Diastolic (Cols 2 to 6)
vital_signs = patients[:, 2:7]
print("\nVital Signs (First 3 patients):\n", vital_signs[:3])

# First 5 patients record using slicing
first_five = patients[:5, :]
print("\nFirst 5 Patients Shape:", first_five.shape)

# Records of selected patients using 2D & Fancy indexing (Patient 102, 107, 109)
selected_patients = patients[[1, 6, 8], :]
print("\nSelected High-Age Patients (102, 107, 109):\n", selected_patients[:, :4])

# Negative indexing: Extract last patient and last column
print("\nLast Patient Record (Negative Index [-1]):\n", patients[-1])
print("Hospital Stays of all patients (Negative Index [:, -1]):", patients[:, -1])
```

## Task 4 – Patient Risk Identification

```python
temp_condition = patients[:, 2] > 38.0
spo2_condition = patients[:, 4] < 95.0
hr_condition   = patients[:, 3] > 100.0
bp_condition   = patients[:, 5] > 140.0

# Combined High-Risk Condition: Any vital sign abnormal
high_risk_mask = temp_condition | spo2_condition | hr_condition | bp_condition

# Identify indices using np.where()
high_risk_indices = np.where(high_risk_mask)[0]
critical_patient_ids = patient_ids[high_risk_indices]

print("--- Patient Risk Diagnostics ---")
print("High Temperature (>38°C) Patient IDs :", patient_ids[temp_condition])
print("Low SpO2 (<95%) Patient IDs          :", patient_ids[spo2_condition])
print("High Heart Rate (>100 bpm) IDs       :", patient_ids[hr_condition])
print("High Systolic BP (>140 mmHg) IDs     :", patient_ids[bp_condition])
print("--------------------------------------------------")
print("PATIENTS REQUIRING IMMEDIATE ATTENTION:", critical_patient_ids)
```

## Task 5 – Element-wise Medical Calculations

```python
# Pulse Pressure = Systolic BP - Diastolic BP
pulse_pressure = patients[:, 5] - patients[:, 6]

# Medication per Day of Hospital Stay (Dose / Stay)
dose_per_day = patients[:, 7] / patients[:, 8]

# Heart Rate Percentage of Max Theoretical HR (220 - Age)
max_hr_theoretical = 220.0 - patients[:, 1]
hr_percentage = (patients[:, 3] / max_hr_theoretical) * 100.0

# Absolute difference from normal temperature (37.0°C)
temp_deviation = np.abs(patients[:, 2] - 37.0)

# Mathematical transformations: Squared, Square Root, Log, Exponential
stay_squared = np.square(patients[:, 8])
dose_sqrt = np.sqrt(patients[:, 7])
log_dose = np.log(patients[:, 7])
exp_temp_norm = np.exp(patients[:, 2] - 37.0)

print("--- Derived Clinical Metrics (First 5 Patients) ---")
print("Patient IDs       :", patient_ids[:5])
print("Pulse Pressure    :", np.round(pulse_pressure[:5], 2))
print("Dose / Day (mg)   :", np.round(dose_per_day[:5], 2))
print("HR % of Max       :", np.round(hr_percentage[:5], 2))
print("Temp Deviation °C :", np.round(temp_deviation[:5], 2))
print("Sqrt(Dose)        :", np.round(dose_sqrt[:5], 2))
```

## Task 6 – Reshaping and Manipulation

```python
# Flattening / 1D Conversion: Flatten vital signs
vitals_flat = vital_signs.ravel()
print("Flattened Vitals Size:", vitals_flat.size)

# Reshaping: Reshape vital signs (10x5) into (5x10)
vitals_reshaped = vital_signs.reshape(5, 10)
print("Reshaped Vitals (5x10) Shape:", vitals_reshaped.shape)

# Transpose Dataset
patients_transposed = patients.T
print("Transposed Dataset Shape (Features x Patients):", patients_transposed.shape)

# Vertical Stacking: Append a new emergency incoming patient record
new_patient = np.array([[111, 50, 38.8, 102, 92, 155, 94, 30, 5]])
combined_patients = np.vstack((patients, new_patient))
print("V-Stacked Dataset Shape (After adding Patient 111):", combined_patients.shape)

# Horizontal Stacking: Append calculated Pulse Pressure as a new column
pulse_col = pulse_pressure.reshape(-1, 1)
patients_with_pp = np.hstack((patients, pulse_col))
print("H-Stacked Dataset Shape (With Pulse Pressure):", patients_with_pp.shape)

# Splitting: Split patients into two monitoring wards (5 patients each)
ward_a, ward_b = np.vsplit(patients, 2)
print("Ward A (First Group) Shape:", ward_a.shape)
print("Ward B (Second Group) Shape:", ward_b.shape)
```

## Task 7 – Statistical Patient Analysis

```python
# Selected columns: Age(1), Temp(2), HR(3), SpO2(4), Systolic(5), Dose(7), Stay(8)
means   = np.mean(patients, axis=0)
stds    = np.std(patients, axis=0)
vars_   = np.var(patients, axis=0)
medians = np.median(patients, axis=0)
mins    = np.amin(patients, axis=0)
maxs    = np.amax(patients, axis=0)

# Percentiles for Heart Rate and Systolic BP (25th, 75th)
hr_p25, hr_p75 = np.percentile(patients[:, 3], [25, 75])
bp_p25, bp_p75 = np.percentile(patients[:, 5], [25, 75])

# Cumulative Metrics
cum_dose = np.cumsum(patients[:, 7])
cum_stay = np.cumsum(patients[:, 8])

print("--- Hospital Statistical Analysis ---")
print(f"Temperature  -> Mean: {means[2]:.2f}°C | Std: {stds[2]:.2f} | Median: {medians[2]:.2f}°C")
print(f"Heart Rate   -> Mean: {means[3]:.1f} bpm | Min: {mins[3]:.0f} | Max: {maxs[3]:.0f} | 25th-75th: [{hr_p25:.1f}, {hr_p75:.1f}]")
print(f"SpO2 (%)     -> Mean: {means[4]:.2f}% | Min: {mins[4]:.0f}% | Median: {medians[4]:.0f}%")
print(f"Systolic BP  -> Mean: {means[5]:.1f} mmHg | Std: {stds[5]:.2f} | Variance: {vars_[5]:.2f}")
print("--------------------------------------------------")
print("Cumulative Medication Dispersed (mg) :", cum_dose)
print("Cumulative Bed Days Occupied (days)  :", cum_stay)
```

## Task 8 – High-Risk and Best-Condition Patients

```python
idx_lowest_spo2 = np.argmin(patients[:, 4])
idx_highest_temp = np.argmax(patients[:, 2])
idx_highest_hr   = np.argmax(patients[:, 3])
idx_longest_stay = np.argmax(patients[:, 8])

print("--- Critical Condition Patient Tracking ---")
print(f"Lowest SpO2 Level       : Patient {int(patients[idx_lowest_spo2, 0])} ({patients[idx_lowest_spo2, 4]}%)")
print(f"Highest Core Temp       : Patient {int(patients[idx_highest_temp, 0])} ({patients[idx_highest_temp, 2]}°C)")
print(f"Highest Heart Rate      : Patient {int(patients[idx_highest_hr, 0])} ({patients[idx_highest_hr, 3]} bpm)")
print(f"Longest Hospital Stay   : Patient {int(patients[idx_longest_stay, 0])} ({patients[idx_longest_stay, 8]:.0f} days)")

# Best condition patient: Lowest temp deviation + highest SpO2
idx_best_spo2 = np.argmax(patients[:, 4])
print(f"Best Oxygen Saturation  : Patient {int(patients[idx_best_spo2, 0])} ({patients[idx_best_spo2, 4]}%)")
```

## Task 9 – Sorting and Searching

```python
# Rank by Hospital Stay (Descending: longest stay to shortest)
stay_rank_idx = np.argsort(patients[:, 8])[::-1]
print("--- Patient Ranking by Hospital Stay (Longest to Shortest) ---")
for rank, idx in enumerate(stay_rank_idx, start=1):
    pid = int(patients[idx, 0])
    stay = patients[idx, 8]
    print(f"Rank {rank}: Patient {pid} ({stay:.0f} days)")

# Sort by SpO2 (Ascending: most critical first)
spo2_sort_idx = np.argsort(patients[:, 4])
print("\n--- Patients Sorted by Oxygen Level (Most Critical First) ---")
print("Patient IDs :", patient_ids[spo2_sort_idx])
print("SpO2 Values :", patients[spo2_sort_idx, 4])
```
