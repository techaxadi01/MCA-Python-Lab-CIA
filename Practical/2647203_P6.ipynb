```python
# Q1: Import the given "UScereal" dataset
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("UScereal.csv")
df.head()
```

```python
print(df.size)
print(df.shape)
```

```python
# Q2: Find summary statistics of the dataset
df.describe()
```

```python
# Q3: Check for missing values
df.isnull().sum()
```

```python
# Q4: Average protein value of each Manufacturer
df.groupby("mfr")["protein"].mean()
```

```python
# Q5: Cereal with highest sugar from Manufacturer G
g_cereals = df[df["mfr"] == "G"]
high_sugar_cereal = g_cereals.loc[g_cereals["sugars"].idxmax()]
print("Cereal Name :", high_sugar_cereal["Name"])
print("Sugar Value :", high_sugar_cereal["sugars"], "g")
```

```python
# Q6: Data Visualizations

# (a) Count of cereals across display shelves (Bar plot)
plt.figure(figsize=(6, 4))
df["shelf"].value_counts().sort_index().plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Distribution of Cereals by Display Shelf")
plt.xlabel("Shelf Level (1: Low, 2: Mid, 3: Top)")
plt.ylabel("Number of Cereals")
plt.show()

# (b) Average fiber content by manufacturer (Bar plot)
plt.figure(figsize=(6, 4))
df.groupby("mfr")["fibre"].mean().plot(kind="bar", color="salmon", edgecolor="black")
plt.title("Average Dietary Fiber by Manufacturer")
plt.xlabel("Manufacturer (G, K, N, P, Q, R)")
plt.ylabel("Mean Dietary Fiber (g)")
plt.show()

# (c) Sodium vs Sugars (Scatter plot)
plt.figure(figsize=(6, 4))
plt.scatter(df["sugars"], df["sodium"], color="purple", alpha=0.7)
plt.title("Relationship between Sugar and Sodium Content")
plt.xlabel("Sugars (g)")
plt.ylabel("Sodium (mg)")
plt.show()
```
