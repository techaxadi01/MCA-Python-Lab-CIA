# String Basics & Character Conversions

```python
# ASCII / Unicode conversions
# chr() converts an ASCII integer to its character
print("chr(97):", chr(97))

# ord() converts a character to its ASCII integer
print("ord('A'):", ord("A"))
```

## String Operations & Slicing

```python
# String initialization
x = "Aditya"
sp = " "
y = "Kumar"

# String concatenation and repetition (* operator)
print("Concatenation:", x + sp + y)
print("Repetition:", (x + sp) * 5)

# String slicing: [start:stop:step] (stop index is excluded)
print("y[:2] (first 2 chars):", y[:2])
print("x[0:10] (stop beyond length):", x[0:10])
print("x[0:10:2] (step of 2):", x[0:10:2])
print("x[0:3] (index 0 to 2):", x[0:3])
print("y[-5:-3] (negative slicing):", y[-5:-3])
```
