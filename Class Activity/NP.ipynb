```python
x = np.array([[1, 2, 3], [4, 5, 6]])
y = np.array([[7, 8, 9], [10, 11, 12]])

print("x : \n", x)
print("y : \n", y)
print()
```

```python
print("x.reshape(3, 2) \n", x.reshape(3, 2), "\n")
print("x.ravel() \n", x.ravel(), "\n")
print("x.flatten() \n", x.flatten(), "\n")
print("x.T \n", x.T, "\n")
print("np.concatenate(([x, y]), axis=0) \n", np.concatenate(([x, y]), axis=0), "\n")
print("np.vstack(([x, y])) \n", np.vstack(([x, y])), "\n")
print("np.hstack(([x, y])) \n", np.hstack(([x, y])), "\n")
print("np.split(x, 2) \n", np.split(x, 2), "\n")
print("np.expand_dims(x, axis=0) \n", np.expand_dims(x, axis=0), "\n")
print("np.squeeze(np.expand_dims(x, axis=0)) \n", np.squeeze(np.expand_dims(x, axis=0)), "\n")
```

```python
print("x : ", x, "\n")
print("x.sum() \n", x.sum(), "\n")
print("np.sum(x, axis=0) \n", np.sum(x, axis=0), "\n")
print("x.mean() \n", x.mean(), "\n")
print("np.mean(x, axis=1) \n", np.mean(x, axis=1), "\n")
print("x.std() \n", x.std(), "\n")
print("x.var() \n", x.var(), "\n")
print("x.min() \n", x.min(), "\n")
print("x.max() \n", x.max(), "\n")
print("x.argmin() \n", x.argmin(), "\n")
print("x.argmax() \n", x.argmax(), "\n")
print("np.median(x) \n", np.median(x), "\n")
print("np.percentile(x, 75) \n", np.percentile(x, 75), "\n")
print("np.cumsum(x) \n", np.cumsum(x), "\n")
print("np.cumprod(x) \n", np.cumprod(x), "\n")
```

```python
p = np.array([[4, 7], [2, 6]])
q = np.array([[1, 2], [3, 4]])

print("p : \n", p)
print("q : \n", q)
print()
```

```python
print("np.dot(p, q) \n", np.dot(p, q), "\n")
print("p @ q \n", p @ q, "\n")
print("np.linalg.inv(p) \n", np.linalg.inv(p), "\n")
print("np.linalg.det(p) \n", np.linalg.det(p), "\n")
print("np.linalg.eig(p) \n", np.linalg.eig(p), "\n")
print("np.linalg.norm(p) \n", np.linalg.norm(p), "\n")
print("np.linalg.solve(p, q) \n", np.linalg.solve(p, q), "\n")
```
