## Compound Interest Rate

John deposited an initial amount of **Rs. 3000** in his account. After **3 years**, the balance reached **Rs. 3335.80** due to compound interest. What is the interest rate per annum?

```python
# A (in CI) = P (1 + R/100)^T
import math

p = 3000
t = 3
a = 3335.8

r = (math.pow(a/p,(1/t)) - 1) * 100

print(f"The rate is = {r:.2f}")
```

## Area of a Triangle

The three sides of a triangle are **3 m**, **4 m**, and **6 m**. Calculate its area using Heron's formula.

```python
a = 3
b = 4
c = 6

s = (a+b+c) / 2

ar = (s * (s-a) * (s-b) * (s-c)) ** 0.5

print("Area is = ", round(ar, 2))
```

## Number Guessing Game

Generate a random number between **1 and 100** and let the user guess it. After each guess, say whether the guess is greater than, equal to, or smaller than the answer. End the game when the guess is correct.

```python
import random

no = random.randint(1, 100)
count = 0

while True:
    guess = int(input("Guess a number between 1 and 100: "))
    count += 1
    if guess > no:
        print("Your guess is greater than the no. Try Again. \n")
    elif guess < no:
        print("Your guess is smaller than the no. Try Again. \n")
    else:
        print("Correct! Game over.")
        break

print(f"\n You guessed the no. {no} in {count} guesses. \n")
```

Ask the user for three integers, `a`, `b`, and `c`, and print the largest value. If the largest value occurs more than once, print `Tie`.

```python
a = int(input("Enter a: "))
b = int(input("Enter b: "))
c = int(input("Enter c: "))

print(max(a, b, c))

if (a == b and a >= c) or (a == c and a >= b) or (b == c and b >= a) or (a == b and b == c):
    print("Tie")
    
```
