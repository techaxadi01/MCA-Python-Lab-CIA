## 1. Creating DataFrames & Tidy Data

```python
import pandas as pd
import numpy as np

# Specify values for each column
df_cols = pd.DataFrame({
    'a': [4, 5, 6],
    'b': [7, 8, 9],
    'c': [10, 11, 12]
}, index=[1, 2, 3])
print("DataFrame specified by columns:\n", df_cols)

# Specify values for each row
df_rows = pd.DataFrame(
    [[4, 7, 10],
     [5, 8, 11],
     [6, 9, 12]],
    index=[1, 2, 3],
    columns=['a', 'b', 'c']
)
print("\nDataFrame specified by rows:\n", df_rows)

# Create DataFrame with MultiIndex
df_multi = pd.DataFrame(
    {'a': [4, 5, 6], 'b': [7, 8, 9], 'c': [10, 11, 12]},
    index=pd.MultiIndex.from_tuples([('d', 1), ('d', 2), ('e', 2)], names=['n', 'v'])
)
print("\nDataFrame with MultiIndex:\n", df_multi)
```

## 2. Reshaping Data & Method Chaining
Gathering/spreading rows, pivoting, sorting, and method chaining.

```python
df_reshape = pd.DataFrame({'a': [4, 5, 6], 'b': [7, 8, 9], 'c': [10, 11, 12]}, index=[1, 2, 3])

# Melt (Gather columns into rows)
melted = pd.melt(df_reshape)
print("Melted DataFrame:\n", melted)

# Pivot (Spread rows into columns)
pivoted = melted.pivot(columns='variable', values='value')
print("\nPivoted DataFrame:\n", pivoted)

# Method Chaining
df_chained = (
    pd.melt(df_reshape)
    .rename(columns={'variable': 'var', 'value': 'val'})
    .query('val >= 8')
)
print("\nMethod Chaining Result:\n", df_chained)

# Concatenation & Sorting
df1 = pd.DataFrame({'a': [1, 2], 'b': [3, 4]})
df2 = pd.DataFrame({'a': [5, 6], 'b': [7, 8]})

concat_rows = pd.concat([df1, df2], axis=0)
concat_cols = pd.concat([df1, df2], axis=1)
sorted_df = concat_rows.sort_values('a', ascending=False)
print("\nSorted DataFrame (High to Low):\n", sorted_df)
```

## 3. Subsetting Observations (Rows) & Variables (Columns)
Filtering data using logic, regular expressions, `loc`, `iloc`, `at`, and `iat`.

```python
df_sub = pd.DataFrame({
    'Length': [6.5, 7.2, 8.1, 5.4, 7.8],
    'Width': [3.1, 2.9, 3.8, 3.0, 2.5],
    'Species': ['setosa', 'versicolor', 'virginica', 'setosa', 'virginica']
})

# Subsetting Rows
print("Length > 7:\n", df_sub[df_sub.Length > 7])
print("\nQuery (Length > 7 and Width < 3.5):\n", df_sub.query('Length > 7 and Width < 3.5'))
print("\nTop 2 Largest Entries:\n", df_sub.nlargest(2, 'Length'))

# Subsetting Columns
print("\nSelected Columns:\n", df_sub[['Length', 'Species']])
print("\nRegex Column Selection (Starts with 'S'):\n", df_sub.filter(regex='^S'))

# Selection using loc, iloc, at, iat
print("\nloc Selection:\n", df_sub.loc[df_sub['Length'] > 7, ['Length', 'Species']])
print("\niloc Selection:\n", df_sub.iloc[0:2, 0:2])
print("\nat Access:\n", df_sub.at[0, 'Species'])
print("\niat Access:\n", df_sub.iat[0, 2])
```

## 4. Grouping & Summarizing Data
Groupby operations, summary statistics, vector calculations, and rolling/expanding windows.

```python
df_group = pd.DataFrame({
    'group': ['A', 'A', 'B', 'B', 'C'],
    'val1': [10, 20, 10, 30, 50],
    'val2': [1, 2, 3, 4, 5]
})

# Summarize
print("Value Counts:\n", df_group['group'].value_counts())
print("\nSummary Description:\n", df_group.describe())

# GroupBy Aggregations
print("\nGroup Mean:\n", df_group.groupby('group').mean(numeric_only=True))
print("\nGroup Aggregations:\n", df_group.groupby('group').agg(['sum', 'max']))

# Vector & Window Functions
df_group['cumsum'] = df_group['val1'].cumsum()
df_group['rolling_mean_2'] = df_group['val1'].rolling(window=2).mean()
print("\nDataFrame with Cumulative Sum & Rolling Mean:\n", df_group)
```

## 5. Combining Data Sets & Handling Missing Data
Standard joins, set operations, filtering joins, and missing value handling.

```python
adf = pd.DataFrame({'x1': ['A', 'B', 'C'], 'x2': [1, 2, 3]})
bdf = pd.DataFrame({'x1': ['A', 'B', 'D'], 'x3': [True, False, True]})

# Merging / Joins
left_join = pd.merge(adf, bdf, how='left', on='x1')
inner_join = pd.merge(adf, bdf, how='inner', on='x1')
outer_join = pd.merge(adf, bdf, how='outer', on='x1')

print("Left Join:\n", left_join)
print("\nOuter Join:\n", outer_join)

# Handling Missing Data
print("\nDrop missing rows:\n", outer_join.dropna())
print("\nFill missing values:\n", outer_join.fillna({'x2': 0, 'x3': False}))

# Filtering Joins
print("\nRows in adf present in bdf:\n", adf[adf.x1.isin(bdf.x1)])
print("\nRows in adf NOT present in bdf:\n", adf[~adf.x1.isin(bdf.x1)])
```

## 6. Plotting, Type Conversion, Datetime & String Operations
Vectorized string methods, datetime accessor (`dt`), mapping, and DataFrame visualization.

```python
# String Operations
s_str = pd.Series(['Apple', 'Banana', 'Cherry', 'Date'])
print("String Lengths:\n", s_str.str.len())
print("\nUppercase Strings:\n", s_str.str.upper())

# Datetime Operations
s_date = pd.to_datetime(pd.Series(['2026-01-15', '2026-06-20', '2026-10-27']))
print("\nExtracted Year:\n", s_date.dt.year)
print("Extracted Quarter:\n", s_date.dt.quarter)

# Type Conversion & Mapping
s_num = pd.Series(['1', '2', '3']).astype(int)
print("\nMapped Values:\n", s_num.map(lambda x: x * 10))

# Plotting Code Example
%matplotlib inline
df_plot = pd.DataFrame({'x': np.arange(10), 'y': np.random.randn(10).cumsum()})
df_plot.plot(x='x', y='y', title="Pandas Line Plot Sample")
```
