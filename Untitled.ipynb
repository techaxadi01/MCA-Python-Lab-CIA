```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configure visual style
sns.set_theme(style='whitegrid')

# Set up a side-by-side subplot figure
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 1. Scatter Plot (Order vs Height)
sns.scatterplot(ax=axes[0], data=df, x='order', y='height(cm)', s=100, color='teal', alpha=0.8)
axes[0].set_title('Presidential Height over Time (by Order)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Presidency Order', fontsize=12)
axes[0].set_ylabel('Height (cm)', fontsize=12)

# 2. Box Plot of Heights
sns.boxplot(ax=axes[1], data=df, y='height(cm)', color='salmon', width=0.4)
axes[1].set_title('Box Plot of US President Heights', fontsize=14, fontweight='bold')
axes[1].set_ylabel('Height (cm)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Ensure dataset is loaded
df = pd.read_csv('/content/iris - iris.csv')

# Define the subsets for each species to resolve the NameError
setosa = df[df['Species'] == 'Iris-setosa']
versicolor = df[df['Species'] == 'Iris-versicolor']
virginica = df[df['Species'] == 'Iris-virginica']

# Plot SepalWidth for all three species
setosa["SepalWidth"].reset_index(drop=True).plot(kind="line", color='r', linewidth=2, label="setosa")
versicolor["SepalWidth"].reset_index(drop=True).plot(kind="line", color='g', linewidth=2, label="versicolor")
virginica["SepalWidth"].reset_index(drop=True).plot(kind="line", color='b', linewidth=2, label="virginica")

plt.title('Line Plot of SepalWidth for All Species')
plt.xlabel('Sample Index (per species)')
plt.ylabel('Sepal Width')
plt.legend(title='Species')
plt.show()
```
