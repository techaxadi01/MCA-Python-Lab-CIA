John deposited an initial amount of Rs. 3000 in his account. After 3 years, the balance reaches Rs. 3335.80 due to compound interest. What is the interest rate per annum?

```python
# A (in CI) = P (1 + R/100)^T
import math

p = 3000
t = 3
a = 3335.8

r = (math.pow(a/p,(1/t)) - 1) * 100

print(f"The rate is = {r:.2f}")
```

The three sides of a triangle are 3, 4, and 6 m respectively. What is its area?

```python
a = 3
b = 4
c = 6

s = (a+b+c) / 2

ar = (s * (s-a) * (s-b) * (s-c)) ** 0.5

print("Area is = ", round(ar, 2))
```

Generate a random number between 1 and 100, and let the user try to guess it.  
Every time the user enters a guess, tell whether the guess is greater than, equal to, or smaller than the answer.  
If the guess is correct, end the game.

```python
import random

no = random.randint(1, 100)
count = 0

while True:
    guess = int(input("Guess a number between 1 and 100: "))
    count += 1
    if guess > no:
        print("Your guess is greater than the no. Try Again. \n")
    elif guess < no:
        print("Your guess is smaller than the no. Try Again. \n")
    else:
        print("Correct! Game over.")
        break

print(f"\n You guessed the no. {no} in {count} guesses. \n")
```
