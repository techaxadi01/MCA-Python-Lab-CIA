john deposited an initial amount of Rs,3000 in his account.After 3 years,the balance reaches rs.3335.8 due to composite interesst.What is the interest rate per annual?

the three sides of a traingle asre 3,4 and 6m,respectively.what is its area?

```python
a = 3
b = 4
c = 6

s = (a+b+c)/2

ar = (s * (s-a) * (s-b) * (s-c))**0.5

print("Area is = ", round(ar, 2))
```
