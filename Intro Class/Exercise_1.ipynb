Write a program that asks for user input and performs  
(a) Celsius to Fahrenheit conversion.   
(b) Fahrenheit to Celsius conversion.   
(c) Meter to foot conversion.   
(d) Foot to meter conversion.   
(e) Acre to square meter conversion.    
(f) Square meter to acre conversion.    
(g) Pound to kilogram conversion.   
(h) Kilogram to pound conversion.   

```python
# (a)
# Formula: F = (C * 9 / 5) + 32
C = float(input("Enter temperature in Celsius: "))
F = (C * 9 / 5) + 32
print("Temperature in Fahrenheit =", F)
```

```python
# (b)
# Formula: C = (F - 32) * 5 / 9
F = float(input("Enter temperature in Fahrenheit: "))
C = (F - 32) * 5 / 9
print("Temperature in Celsius =", C)
```

```python
# (c)
# Formula: ft = m * 3.28084
m = float(input("Enter length in meters: "))
ft = m * 3.28084
print("Length in feet =", ft)
```

```python
# (d)
# Formula: m = ft * 0.3048
ft = float(input("Enter length in feet: "))
m = ft * 0.3048
print("Length in meters =", m)
```

```python
# (e)
# Formula: m2 = ac * 4046.8564224
ac = float(input("Enter area in acres: "))
m2 = ac * 4046.8564224
print("Area in square meters =", m2)
```

```python
# (f)
# Formula: ac = m2 / 4046.8564224
m2 = float(input("Enter area in square meters: "))
ac = m2 / 4046.8564224
print("Area in acres =", ac)
```

```python
# (g)
# Formula: kg = lb * 0.45359237
lb = float(input("Enter weight in pounds: "))
kg = lb * 0.45359237
print("Weight in kilograms =", kg)
```

```python
# (h)
# Formula: lb = kg / 0.45359237
kg = float(input("Enter weight in kilograms: "))
lb = kg / 0.45359237
print("Weight in pounds =", lb)
```
